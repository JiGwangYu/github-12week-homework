```python
import requests
from bs4 import BeautifulSoup
import pandas as pd

url = "https://www.saramin.co.kr/zf_user/search?search_area=main&search_done=y&search_optional_item=n&searchType=search&searchword=데이터분석"
headers = {"User-Agent": "Mozilla/5.0"}
resp = requests.get(url, headers=headers, timeout=20)
resp.raise_for_status()
soup = BeautifulSoup(resp.text, "html.parser")

BASE = "https://www.saramin.co.kr"
records = []

section = soup.select_one('section.section_search#recruit_info')

companies = [a.get_text(strip=True) for a in section.select('strong.corp_name a')]
titles = [a.get_text(strip=True) for a in section.select('h2.job_tit a')]
links = [
    a["href"] if a["href"].startswith("http") else BASE + a["href"]
    for a in section.select("h2.job_tit a") if a.has_attr("href")
]

conditions_all = []
for cond_div in section.select("div.job_condition"):
    cond_list = []
    for span in cond_div.find_all("span"):
        if span.find("a"):
            txt = " ".join(a.get_text(strip=True) for a in span.find_all("a"))
        else:
            txt = span.get_text(strip=True)
        if txt:
            cond_list.append(txt)
    conditions_all.append(cond_list)

for comp, tit, conds, link in zip(companies, titles, conditions_all, links):
    records.append({
        "Site": "Saramin",
        "Col_Company": comp,
        "Col_Recruit": tit,
        "Col_detail": conds,
        "Col_url": link
    })

df = pd.DataFrame(records)
display(df)
```

```python
df.to_csv("data_tmp/data_saramin.csv", index=False, encoding="utf-8-sig")
```
